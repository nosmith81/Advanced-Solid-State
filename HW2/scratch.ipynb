```python
from sympy import Matrix, symbols, solve, Eq, cos, Rational
import sympy as smp

ep1, ep2 = symbols('\epsilon_1 \epsilon_2')

eta11, eta12, eta22 = symbols('\eta_11 \eta_12 \eta_22')

t11, t12, t22 = symbols('t_{11} t_{12} t_{22}')

Ek = symbols('E_k')

a, b = symbols(['a', 'b'])

a, kx, ky = symbols('a k_x k_y', positive=True)
k = Matrix([kx, ky])
```

```python
from sympy import Matrix, I, exp, Function

# Unevaluated symbolic functions
eta_nm = Function(r'\eta_{nm}')
t_nm   = Function(r't_{nm}')

kx, ky = symbols(['k_x', 'k_y'])

k = Matrix([kx, ky])

def get_etak(Rj_lst, taudiff, k):
    return sum(eta_nm((R + taudiff).norm()) * exp(I * k.dot(R)) for R in Rj_lst)

def get_tk(Rj_lst, taudiff, k):
    return sum(t_nm((R + taudiff).norm()) * exp(I * k.dot(R)) for R in Rj_lst)


```

### Square Lattice

```python

taudiff = Matrix([0, 0])

# Nearest-Neighbor vectors for square lattice
R_NN = [Matrix([a, 0]), Matrix([-a, 0]), Matrix([0, a]), Matrix([0, -a])]

eta_k = get_etak(R_NN, taudiff, k)

eta_k.simplify(deep = True).rewrite(cos).expand()
```

```python
taudiff = Matrix([0, 0])

R_NNN = [Matrix([a, a]), Matrix([-a, a]), Matrix([a, -a]), Matrix([-a, -a])]

eta_k = get_etak(R_NNN, taudiff, k)

eta_k.simplify(deep = True).rewrite(cos).expand().simplify()
```

### Rectangular Lattice

```python
### Rectangular lattice NN site 1
#b = Rational(0.5) * a

taudiff =  Matrix([0, 0]) 

R_NN = [Matrix([a, 0]), Matrix([-a, 0]), Matrix([0, b]), Matrix([0, -b])]

eta_k = get_etak(R_NN, taudiff, k)

eta_k.simplify(deep = True).rewrite(cos).expand()
```

```python
### Rectangular lattice NNN

R_NNN = [Matrix([a, b]), Matrix([-a, b]), Matrix([-a, -b]), Matrix([a, -b])]

eta_k = get_etak(R_NNN, taudiff, k)

eta_k
```
