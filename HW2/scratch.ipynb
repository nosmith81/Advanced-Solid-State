```python
from sympy import Matrix, symbols, solve, Eq

ep1, ep2 = symbols('\epsilon_1 \epsilon_2')

eta11, eta12, eta22 = symbols('\eta_11 \eta_12 \eta_22')

t11, t12, t22 = symbols('t_{11} t_{12} t_{22}')

Ek = symbols('E_k')
```

```python
etabar = Matrix([
    [eta11, eta12],
    [eta12, eta22]
])

epbar = Matrix([
    [0, 0],
    [0, ep2]
])

tbar = Matrix([
    [t11, t12],
    [t12, t22]
])

M = epbar + tbar @ etabar.inv()
```

```python
P, D = M.diagonalize()
```

```python
D[0,0].simplify(deep = True)
```

```python
D[1,1].simplify(deep = True)
```
